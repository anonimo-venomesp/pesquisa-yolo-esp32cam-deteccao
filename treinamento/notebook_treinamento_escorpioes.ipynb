# Treinamento do Modelo YOLOv8 para Detecção de Escorpiões
Este notebook reproduz o treinamento do modelo YOLOv8n utilizando o dataset de Escorpiões hospedado no Roboflow.
As métricas de validação esperadas ao final do treinamento de 50 épocas (mAP@50 de 92.2% e Precision de 93.4%) estão detalhadas na Tabela VIII do artigo.

```python
from roboflow import Roboflow

# Substitua pela sua chave da API do Roboflow
rf = Roboflow(api_key="COLOQUE_SUA_CHAVE_AQUI")
project = rf.workspace("SEU_WORKSPACE_AQUI").project("scorpion-h2wfc-jmwsg")
version = project.version(1)
dataset = version.download("yolov8")

print("\n\u2705 Dataset de Escorpiões baixado com sucesso via Roboflow!")
```

```python
import os
from pathlib import Path
from ultralytics import YOLO

# 1. Desativa o Weights & Biases (Opcional)
os.environ["WANDB_DISABLED"] = "true"

# 2. Inicia o treinamento oficial com seed=42 e 50 épocas
pasta_resultados = Path.cwd() / "resultados_treino"
model = YOLO("yolov8n.pt")

print("\n\ud83d\ude80 INICIANDO TREINO DE 50 ÉPOCAS (SEED 42) PARA ESCORPIÕES NA GPU...")
results = model.train(
    data=f"{dataset.location}/data.yaml",
    seed=42,
    epochs=50,
    project=str(pasta_resultados),
    name="yolov8_escorpioes",
    exist_ok=True,
)
```

```python
# 3. Exibe a tabela comparativa ao término do treinamento
print("\n" + "=" * 70)
print("\ud83c\udfaf COMPARATIVO DO TREINO LOCAL COM AS MÉTRICAS DO ARTIGO (TABELA VIII) \ud83c\udfaf")
print("=" * 70)
print(f"Precisão (P):    {results.box.mp:.4f}   | Referência Artigo: ~0.9340 (93.4%)")
print(f"Revocação (R):   {results.box.mr:.4f}   | Referência Artigo: ~0.9050 (90.5%)")
print(f"mAP50 (B):       {results.box.map50:.4f}  | Referência Artigo: ~0.9220 (92.2%)")
print("=" * 70)
```
